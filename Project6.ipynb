```python
import statsmodels.api as sm
```

```python
# Create log price
spx["Log_Close"] = np.log(spx["Adj Close"])

# Create time index
spx["Time_Index"] = np.arange(len(spx))

# Regression model
X = sm.add_constant(spx["Time_Index"])
y = spx["Log_Close"]

model = sm.OLS(y, X).fit()
print(model.summary())
```

### Regression Analysis: Long-Term Trend in S&P 500 Prices

To quantify the long-term growth trend of the S&P 500, I fit an OLS regression model using
the natural logarithm of the adjusted closing price (`Log_Close`) as the dependent variable and a
time index as the independent variable. The log transformation linearizes the exponential growth
pattern seen in financial markets and is appropriate for modeling trend over long horizons.

The regression produced an **R-squared of 0.959**, meaning that **over 95% of the variation in
log-transformed S&P 500 prices is explained by time alone**. This confirms a very strong long-run upward trend.

The coefficient for `Time_Index` is **0.0003** (p < 0.001), indicating a statistically significant
increase in the index over time. Because this coefficient reflects the average daily log-growth,
it implies a positive and persistent long-term growth rate for the index. The intercept (1.8047)
represents the estimated starting log-price at the beginning of the series.

The extremely low p-values (< 0.001) for both the time coefficient and the overall F-statistic
(5.39 × 10⁵) indicate that the model is statistically significant and that time is a strong
predictor of long-term price movement. The fitted regression line confirms that the S&P 500’s
growth trajectory is well-captured by an exponential trend, which supports my research question
about long-term growth behavior in the index.


```python
from scipy.stats import f_oneway

decade_groups = [group["Adj Close"].values 
                 for _, group in spx.groupby("Decade")]

f_stat, p_value = f_oneway(*decade_groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

### ANOVA Test: Differences in S&P 500 Prices Across Decades

To test whether S&P 500 price levels differ significantly across decades, I conducted a
one-way ANOVA using the adjusted closing price grouped by decade. This analysis allows me to
examine whether long-term structural differences exist over different historical periods.

The ANOVA yielded an **F-statistic of 20,374.85** with a **p-value of 0**. Because the
p-value is far below the 0.05 significance threshold, I reject the null hypothesis that all
decades have the same mean price level.

This result indicates **highly significant differences in average S&P 500 prices across decades**.
This is consistent with the substantial long-term growth observed in the data: early decades
(1920s–1950s) had much lower price levels, while more recent decades (2000s–2020s) exhibit
dramatically higher averages and greater volatility.

Overall, the ANOVA confirms that long-term market growth is not uniform over time and that
decade-level differences are statistically meaningful. These findings reinforce the results
from the regression analysis, demonstrating that growth in the S&P 500 is both strong and
structurally distinct across historical eras.
